4. Connecting multiple inputs to same output 

```python
import time

mux.reset()
time.sleep(0.5)
output_channel = 1
input_pins = [15, 26, 67, 78]

for pin in input_pins:
    mux.switch(pin, output_channel, 1)
    time.sleep(0.2)

mux.full_status(plot=True)
```

5. Connecting multiple outputs to a same input

```python
import time

mux.reset()
mux.full_status(plot=True)

time.sleep(0.2)
output_channel = [1, 3, 4]
input_pin = 3

for channels in output_channel:
    mux.switch(input_pin, channels, 1)
    time.sleep(0.2)

mux.full_status(plot=True)
```
